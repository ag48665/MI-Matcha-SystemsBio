```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()

DATA_DIR = PROJECT_ROOT / "data" / "raw"

compounds = pd.read_csv(DATA_DIR / "matcha_compounds.csv")

compounds
```

```python
print(compounds.info())
print("\nMissing values:")
print(compounds.isna().sum())

print("\nCompound groups:")
print(compounds["compound_group"].value_counts())
```

```python
import matplotlib.pyplot as plt

group_counts = compounds["compound_group"].value_counts()

ax = group_counts.plot(kind="bar")

ax.set_title("Bioactive compound groups in matcha")
ax.set_xlabel("Compound group")
ax.set_ylabel("Number of compounds")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
OUTPUT_DIR = PROJECT_ROOT / "outputs" / "figures"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

ax = group_counts.plot(kind="bar")

ax.set_title("Bioactive compound groups in matcha")
ax.set_xlabel("Compound group")
ax.set_ylabel("Number of compounds")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "compound_group_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
